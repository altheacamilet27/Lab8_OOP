```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.__balance = balance
        self.account_holder = account_holder

    def deposit(self, amount):
        self.__balance += amount
        return f"New Balance: {self.__balance}"

    def get_balance(self):
        return f"Balance: {self.__balance}"

account = BankAccount("John Doe", 1000)
print(account.deposit(500))
print(account.get_balance())
```

```python
class Box:
    def __init__(self, weight):
        self.weight = weight

    def __add__(self, other):
        return Box(self.weight + other.weight)  # ✅ return object

box1 = Box(5)
box2 = Box(10)

box3 = box1 + box2
print(f"Total weight: {box3.weight} kg")
```

```python
from abc import ABC, abstractmethod

class Appliance(ABC):
    @abstractmethod
    def turn_on(self):
        pass

class Fan(Appliance):
    def turn_on(self):
        return "Fan is spinning"

fan = Fan()
print(fan.turn_on())
```

```python
class ATM:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            return "Insufficient balance"
        self.balance -= amount
        return f"Withdrawn {amount}, Remaining balance: {self.balance}"

atm = ATM(1000)
print(atm.withdraw(200))
```

```python
# =========================
# LIBRARY BOOK SYSTEM
# =========================

class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.available = True

class Library:
    def __init__(self):
        self.books = []

    def add_book(self, title, author):
        self.books.append(Book(title, author))

    def borrow_book(self, title):
        for book in self.books:
            if book.title == title:
                if book.available:
                    book.available = False
                    return "Book borrowed successfully"
                return "Book already borrowed"
        return "Book not found"

    def return_book(self, title):
        for book in self.books:
            if book.title == title:
                book.available = True
                return "Book returned successfully"
        return "Book not found"

    def display(self):
        print("\n--- Library Books ---")
        for book in self.books:
            status = "Available" if book.available else "Borrowed"
            print(f"{book.title} by {book.author} - {status}")

# TEST RUN
lib = Library()
lib.add_book("Python 101", "John Doe")
lib.add_book("OOP Basics", "Jane Smith")

print(lib.borrow_book("Python 101"))
lib.display()
print(lib.return_book("Python 101"))
lib.display()
```

```python
# =========================
# PAYROLL SYSTEM
# =========================

class Employee:
    def __init__(self, name, hourly_rate, hours_worked):
        self.name = name
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked

    def calculate_salary(self):
        overtime = 0

        if self.hours_worked > 160:
            overtime_hours = self.hours_worked - 160
            overtime = overtime_hours * (self.hourly_rate * 1.5)
            base = 160 * self.hourly_rate
        else:
            base = self.hours_worked * self.hourly_rate

        return base, overtime

# TEST RUN
emp = Employee("Michael", 100, 170)
base, overtime = emp.calculate_salary()

print("\n--- Payroll Details ---")
print("Employee Name:", emp.name)
print("Hours Worked:", emp.hours_worked)
print("Base Salary:", base)
print("Overtime Pay:", overtime)
```

```python
# =========================
# SHOPPING CART SYSTEM
# =========================

class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

class Cart:
    def __init__(self):
        self.items = []

    def add_to_cart(self, product):
        if product.stock > 0:
            self.items.append(product)
            product.stock -= 1
            return f"{product.name} added to cart"
        return "Out of stock"

    def total_price(self):
        return sum(item.price for item in self.items)

# TEST RUN
p1 = Product("Shoes", 1000, 5)
p2 = Product("Shirt", 500, 2)

cart = Cart()

print("\n--- Cart Actions ---")
print(cart.add_to_cart(p1))
print(cart.add_to_cart(p2))

print("\n--- Cart Summary ---")
print("Total Price:", cart.total_price())
```
